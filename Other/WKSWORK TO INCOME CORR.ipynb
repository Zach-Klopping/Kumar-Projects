```python
#use 3.12.7
import pandas as pd
import plotly.express as px
import numpy as np
from scipy import stats
file1= open("WKS WORK TO INCOME TABLE.csv", "r")
df = pd.read_table("WKS WORK TO INCOME TABLE.csv", sep='\\,', engine='python')
df.rename(columns={'"INCWAGE"': 'INCWAGE', '"WKSWORK1"': 'WKSWORK'}, inplace=True)
df.sort_values('INCWAGE', ascending=False, inplace=True)
df1= df[(np.abs(stats.zscore(df)) < 3).all(axis=1)]
fig=px.scatter(df1,x='WKSWORK', y='INCWAGE', title="Correlation of Weeks Worked to Income", trendline="ols", trendline_color_override="blue",color_discrete_sequence=['green'])
fig.update_layout(xaxis_title="Weeks Worked (Previous Year)", yaxis_title="Pre-Tax Salary & Wage Income")
df1
```
